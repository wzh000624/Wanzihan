```python
# 导入 PyTorch 库
# → 用于检查是否可以使用 GPU (CUDA)
import torch

# 导入 Hugging Face Transformers 的 pipeline 函数
# → 该函数可以将复杂的模型处理过程简化
from transformers import pipeline


# 创建情感分析 (Sentiment Analysis) 管道
classifier = pipeline(

    # 指定要执行的 NLP 任务
    # sentiment-analysis: 将句子的情感分类为正面或负面
    task="sentiment-analysis",

    # 指定要使用的预训练模型
    # 使用 DistilBERT 模型，该模型已在 SST-2 情感分析数据集上进行了微调
    model="distilbert-base-uncased-finetuned-sst-2-english",

    # 指定运行设备
    # 如果有 GPU (CUDA)，则使用第一个 GPU (device=0)
    # 如果没有 GPU，则使用 CPU (device=-1)
    device=0 if torch.cuda.is_available() else -1
)
```

```python
# 两个用于情感分析的英文句子
texts = [
    "This lecture is absolutely fascinating!",  # 正面句子
    "I couldn't understand anything at all."    # 负面句子
]

# 对多个句子同时进行情感分析
results = classifier(texts)

# 逐一输出每个句子及其分析结果
for text, result in zip(texts, results):

    # label : 情感分析结果 (POSITIVE 或 NEGATIVE)
    # score : 模型对结果的置信度分数
    # :.4f  : 将分数输出到小数点后4位
    print(f"{result['label']:8s} {result['score']:.4f} <- {text}")
```

```python
# 从 Hugging Face 自动加载分词器的类
from transformers import AutoTokenizer

# 加载能够处理多种语言的多语言 BERT 分词器
tok = AutoTokenizer.from_pretrained(
    "google-bert/bert-base-multilingual-cased"
)

# 准备英文和韩文示例句子
samples = {
    "英语": "Natural language processing is fascinating.",
    "한국어": "자연어처리는 정말 흥미롭습니다.",
    "중국어": "自然语言处理真的很有趣。"
}

# 逐一处理英文、韩文和中文句子
for lang, text in samples.items():

    # 将句子分成 BERT 可以处理的小单元，即 token
    tokens = tok.tokenize(text)

    # 将每个 token 转换为模型使用的数字 ID
    ids = tok.convert_tokens_to_ids(tokens)

    # 输出使用的语言
    print(f"[{lang}]")

    # 输出原始句子
    print("원문:", text)

    # 输出分词结果
    print("토큰:", tokens)

    # 输出句子被分成多少个 token
    print("토큰 수:", len(tokens))

    # 输出每个 token 对应的数字 ID
    print("ID:", ids)

    # 在英文和韩文结果之间添加空行
    print()
```

```python
# 一个函数，用于对单个句子进行分词并计算简单的统计信息
def token_stats(tokenizer, text):

    # 将句子分成 token
    tokens = tokenizer.tokenize(text)

    # 将计算结果以字典形式返回
    return {
        # 原始文本中包含的总字符数
        "chars": len(text),

        # 分词后的总 token 数
        "tokens": len(tokens),

        # 计算每个 token 平均由多少个字符组成
        "chars_per_token": len(text) / max(len(tokens), 1)
    }


# 逐一处理前面创建的英文、韩文和中文示例句子
for lang, text in samples.items():

    # 输出每种语言的 token 统计结果
    print(lang, token_stats(tok, text))
```
